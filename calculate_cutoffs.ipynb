```python
from src.carbon_intensity import CarbonIntensity
import numpy as np
```

```python
carbonIntensity = CarbonIntensity(green_win_length=24, normalize=False)
```

```python
results = {}

for split in ["validation","test"]: 
    results[split] = {}

    carbonIntensity.set_mode(split)
    start_slot = carbonIntensity.window_start_slot
    end_slot = start_slot + carbonIntensity.window_slot_count

    data_slice = carbonIntensity.carbonIntensityList[start_slot:end_slot-1] 

    results[split][10] = np.percentile(data_slice,10)
    results[split][25] = np.percentile(data_slice,25)
    results[split][50] = np.percentile(data_slice,50)

    results[split][60] = np.percentile(data_slice,60)
    results[split][70] = np.percentile(data_slice,70)
    results[split][80] = np.percentile(data_slice,80)
    results[split][90] = np.percentile(data_slice,90)
    results[split][95] = np.percentile(data_slice,95)
    results[split][97] = np.percentile(data_slice,97)
```
